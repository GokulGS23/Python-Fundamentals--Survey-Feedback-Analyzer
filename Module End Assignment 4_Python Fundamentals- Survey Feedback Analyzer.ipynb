### Step 3: Text Cleaning

```python
import string

def clean_feedback(text):
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = " ".join(text.split())
    return text.lower()

feedback_data["Feedback"] = [
    clean_feedback(text) for text in feedback_data["Feedback"]
]

print(feedback_data["Feedback"])
```

### Step 4: Word Count Insights

```python
def count_word_in_feedbacks(word):
    count = 0

    for feedback in feedback_data["Feedback"]:
        if word.lower() in feedback.split():
            count += 1

    return count

word = input("Enter a word to search: ")
result = count_word_in_feedbacks(word)

print("Number of feedbacks containing the word:", result)
```

### Step 5: Final Summary & Insights

```python
total_feedbacks = len(feedback_data["Feedback"])
average_rating = sum(feedback_data["Rating"]) / len(feedback_data["Rating"])

longest_feedback = max(
    feedback_data["Feedback"],
    key=lambda text: len(text.split())
)

all_words = []

for feedback in feedback_data["Feedback"]:
    all_words.extend(feedback.split())

unique_words = set(all_words)

print("Total feedbacks:", total_feedbacks)
print("Average rating:", round(average_rating, 2))
print("Longest feedback:", longest_feedback)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```
